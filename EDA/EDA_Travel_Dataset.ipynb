```python
# Data is either categorical or numerical, numerical can be either discrete and continuos
# discrete data through its numerical but should be treated as categorical to do the analysis
# but for model building all the categorical data should be ci=onverted to number
```

```python
cat_features=x.select_dtypes(include='object').columns
num_features=x.select_dtypes(exclude='object').columns
```

```python
num_features
```

```python
# scaling -- numerical features
# Categorical -- Onehot encoding
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler # boyh are the data transformation 
from sklearn.compose import ColumnTransformer

# columnTransormer applies transformations to column of an array or pandas dataframe

# This estimator allows different columns or column subsets of the
# input to be transformed separetly and the features generated by each
# transformer will be concatenated to form a single feature space.
# This is useful  for heterogeneous or columar data, to combine several
# feature
# extraction mechanisms or transformation into a single transformer
```

```python
numeric_transformer=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')
preprocessor=ColumnTransformer(
    [
        ('OneHotEncoder', oh_transformer,cat_features),
        ('StandardScaler',numeric_transformer, num_features )
    ])
```

```python
preprocessor
```

```python
x_train=preprocessor.fit_transform(x_train)
pd.DataFrame(x_train)
```
