```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Generate same synthecic datapoint or reproducibility of results
np.random.seed(1)
```

```python
# Generate synthtic x

x=2*np.random.rand(100,1)
```

```python
x
```

```python
# Generate synthtic y

y= 4 + 3*x+1.5*x**2 + np.random.rand(100,1)
```

```python
y
```

### Train test split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.25,random_state=1)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

### perform Polynomial Regression 

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_featr=PolynomialFeatures(degree=2, include_bias=False)
poly_featr
```

```python
x_poly_train=poly_featr.fit_transform(x_train)
x_poly_train
```

```python
from sklearn.linear_model import LinearRegression
poly_reg=LinearRegression()
poly_reg
```

```python
poly_reg.fit(x_poly_train, y_train)
```

```python
poly_reg.coef_
```

```python
poly_reg.intercept_
```

```python
# Make prediction on train data itself for the evaluation metrics
y_poly_predict=poly_reg.predict(x_poly_train)
```

```python
y_poly_predict
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_train, y_poly_predict)
```

```python
plt.scatter(x_train,y_train,label='train data')
```

```python
degree=2
```

```python
x_range=np.linspace(0, 2, 100).reshape(-1, 1)
x_range_poly=poly_featr.transform(x_range)
plt.plot(x_range,poly_reg.predict(x_range_poly), color='red', label=f'Polynomial Regression(Degree{degree})')
plt.xlabel('X')
plt.ylabel('Y')
plt.title("Polynomial regression")
plt.legend()
plt.show()
```
