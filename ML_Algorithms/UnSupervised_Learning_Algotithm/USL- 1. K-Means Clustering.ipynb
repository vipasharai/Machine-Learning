# K-Means Clustering

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_blobs   # help to create cluster
x,y=make_blobs(n_samples=1000, centers=3, n_features=2, random_state=1)
```

```python
x
```

```python
# y  # In Unsupervised we don't use Y
```

```python
plt.scatter(x[:, 0],x[:,1])
```

```python
plt.scatter(x[:, 0],x[:,1], c=y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.3, random_state=1)
```

```python
x_train.shape, x_test.shape
```

```python
from sklearn.cluster import KMeans
Kmeans=KMeans(n_clusters=3) # change value 
Kmeans
```

```python
Kmeans.fit(x_train) # fit_predict also use
```

```python
y_labels=Kmeans.predict(x_train)
y_labels
```

```python
plt.scatter(x_train[:, 0],x_train[:,1], c=y_labels)
```

```python
# How better this clustering model is?
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(x_train, KMeans.labels_)  # close to 1 we have good model
```

```python
Kmeans.labels_
```

### WCSS

```python
Kmeans.inertia_  # K=3  , this gives WCSS distance
```

```python
# Howto select k(centroid) >> elbow method
```

```python
wcss=[]
for k  in range(1, 10):
    Kmeans=KMeans(n_clusters=k)
    Kmeans.fit(x_train)
    wcss.append(Kmeans.inertia_)

wcss  #  wcss is decresing as k increase
```

```python
# plot the elbow curve
plt.plot(range(1, 10), wcss)
plt.xticks(range(1,10))
plt.xlabel('No. of students')
plt.ylabel('WCSS')
plt.show()
```

```python
# Based on the plot abobve the optimal no. of k/ centroid shold be  3
```
