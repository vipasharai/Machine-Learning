# DBSCAN 

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons  # create non-linear

x,y=make_moons(n_samples=250, noise=0.10)
```

```python
x
```

```python
y # will not used in modelling but use in plot
```

```python
plt.scatter(x[:,0], x[:,1], c=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
x_scaled
```

```python
dbscan=DBSCAN(eps=0.3)  # change the value of eps to show the result
dbscan
```

```python
dbscan.fit(x_scaled)
```

```python
dbscan.labels_
```

```python
plt.scatter(x[:,0],x[:, 1], c=dbscan.labels_)
```
