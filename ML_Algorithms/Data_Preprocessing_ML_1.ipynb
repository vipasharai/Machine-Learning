# **Class Imbalance**
 when one class has higher percentage --
 handle:
** 1.Upsampling
 2.downsampling
 3.SMOTE
**


```python
import pandas as pd
import numpy as np

np.random.seed(1) # for reproducibility of result
no_samples=1000
class_0_ratio=0.9
no_class_0 =int(no_samples*class_0_ratio)
no_class_1=100
```

```python
no_class_0, no_class_1
```

```python
class_0={'feature1' : np.random.normal(0,1,no_class_0),
         'feature2' : np.random.normal(0,1,no_class_0),
         'target' : [0]*no_class_0 }
```

```python
class_0=pd.DataFrame(class_0)
```

```python
class_0
```

```python
class_1=pd.DataFrame({'feature1' : np.random.normal(3,1,no_class_1),
         'feature2' : np.random.normal(0,1,no_class_1),
         'target' : [1]*no_class_1 })
```

```python
class_1
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df
```

```python
df.target.value_counts()
```

# Upsampling

```python
df_minority = df[df.target==1]
df_majority = df[df.target==0]
```

```python
df_majority
```

```python
df_minority
```

```python
# Replace >> if replace is true, it means data point can be repeated
from sklearn.utils import resample
df_minority_upsampled=resample(df_minority, replace=True, n_samples= len(df_majority))
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled
```

```python
df_upsampled=pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled
```

```python
df_upsampled.target.value_counts()
```

## DownSampling

```python
df_minority
```

```python
df_majority
```

```python
df_majority_downsampled=resample(df_majority, replace=False, n_samples=len(df_minority),random_state=1)
df_majority_downsampled
```

## OUTLIER

```python
## Outliers are extreme values
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
salary=[11,40,45,68,65,68,78,90,57,74,91,92,88,68,57,48,99,101,68,77,110,140]
salary
```

```python
len(salary)
```

```python
# Five Points summary
#  min, 25th, 50th, 75th percentile 100th percentile
```

```python
np.quantile(salary,[0,0.25,0.50,0.75,1])
```

```python
df=pd.DataFrame(salary,columns=['salary'])
```

```python
df.describe()
```

```python
# To check outliers
# - distplot, boxplot
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df['salary'],kde=True)
plt.title("Dist Plot")

plt.subplot(1,2,2)
sns.boxplot(data=df, x='salary')
plt.title("Box Plot")
plt.show()
```

```python
# here the distribution is not skewed as the outliers are present on both side
# In boxplot , we see there are outliers

# dropping the outlier
# capping the outlier
# replace with mean and median
```

```python
# Drop the Outliers
```

```python
Q1=df["salary"].quantile(0.25)
Q3=df["salary"].quantile(0.75)

IQR=Q3-Q1

lower_fence=Q1-1.5*IQR
upper_fence=Q3+1.5*IQR
```

```python
df_filtered=df[(df.salary >= lower_fence) & (df.salary <= upper_fence)]
```

```python
df.shape
```

```python
df_filtered.shape
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df_filtered['salary'],kde=True)
plt.title("Dist Plot")

plt.subplot(1,2,2)
sns.boxplot(data=df_filtered, x='salary')
plt.title("Box Plot")
plt.show()
```

```python
# Imputation with mean and median
df['salary_imputed_mean']=np.where((df.salary >= upper_fence) | (df.salary <= lower_fence), df['salary'].mean(), df['salary'])
```

```python
df
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df['salary_imputed_mean'],kde=True)
plt.title("Dist Plot")

plt.subplot(1,2,2)
sns.boxplot(data=df, x='salary_imputed_mean')
plt.title("Box Plot")
plt.show()
```

```python
# imputation with median
df['salary_imputed_median']=np.where((df.salary >= upper_fence) | (df.salary <= lower_fence), df['salary'].median(), df['salary'])
```

```python
df
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df['salary_imputed_median'],kde=True)
plt.title("Dist Plot")

plt.subplot(1,2,2)
sns.boxplot(data=df, x='salary_imputed_median')
plt.title("Box Plot")
plt.show()
```

```python
# Capping -- replacing the nearest value which is not outlier
```

```python
df
```

```python
lower_cap = df['salary'].quantile(0.05) # lower cap is 5th percentile
upper_cap = df['salary'].quantile(0.95) # upper cap is 95th percentile
```

```python
lower_cap
```

```python
upper_cap
```

```python
# replace outlier with cap
# -- where(condition, [x,y], /)
# -- return elements chosen from 'x' or 'y' on 'condition'

# if df['salar'] < lower_cap, satisfied then lower_cap and df['salary']>upper_cap, satisfied then upper_cap, df['salary']
# otherwise keep salary  as its is
```

```python
df['salary_cap']=np.where(df['salary'] < lower_cap, lower_cap,
         np.where(df['salary']>upper_cap, upper_cap, df['salary']))
```

```python
df
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df['salary_cap'],kde=True)
plt.title("Dist Plot")

plt.subplot(1,2,2)
sns.boxplot(data=df, x='salary_cap')
plt.title("Box Plot")
plt.show()
```
