```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings ("ignore")
```

```python
df=sns.load_dataset('mpg')
df
```

```python
df.drop("name",axis=1, inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
#  Since we have not done outlier treatment, then better idea be to replace the median value
```

```python
df['horsepower'].median()
```

```python
df['horsepower']=df['horsepower'].fillna(df['horsepower'].median())
df['horsepower']
```

```python
df.isna().sum()
```

```python
df['origin'].value_counts()
```

### Data encoding

```python
df["origin"]=df["origin"].map({"usa":1,"japan":2,"europe":3})
```

```python
df["origin"]=df["origin"].astype(int)
df["origin"]
```

```python
# Seprtae into x and y
```

```python
x=df.drop('mpg',axis=1)
y=df['mpg']
```

```python
x
```

```python
y
```

### train test split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.3, random_state=1)
```

```python
x_train.shape
```

### Simple Linear Regression

```python
from sklearn.linear_model import LinearRegression
Regression_model=LinearRegression()
```

```python
Regression_model
```

```python
Regression_model.fit(x_train, y_train)
```

```python
Regression_model.coef_
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"The coefficent for {col_name} is {Regression_model.coef_[i]}")
```

```python
## Onservation
# Coefficients ar relatively smaller, if one independent variable
# there will be not much difference in preediction
# This is sometime called as smoother model

# This feature might not be contributing in model training
```

```python
from sklearn.metrics import r2_score

y_pred_linear=Regression_model.predict(x_test)
r2_linear=r2_score(y_test, y_pred_linear)
print(f"The r squre of linear regression {r2_linear}")
```

### Regularised model

##  Ridge Regression

```python
from sklearn.linear_model import Ridge
ridge_regression_model=Ridge(alpha=0.1)
ridge_regression_model
# In practical implemantation lambda is alpha
```

```python
ridge_regression_model.fit(x_train, y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"The coefficent for {col_name} is {ridge_regression_model.coef_[i]}")
```

```python
# for ridge regression evaluation

y_pred_ridge=ridge_regression_model.predict(x_test)
r2_ridge=r2_score(y_test, y_pred_ridge)
print(f"The r squre of ridge regression {r2_ridge}")
```

```python
#We dont see much variation in coeff of ridge regression as compare to linear regression
```

# 2 - Hyperparmeter tuning

```python
# Cv with hyperparameter tuning
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import Lasso, Ridge
```

```python
# Define the model
lasso=Lasso()
```

```python
lasso
```

```python
# grid seach cv>> Dictionary with parameter names(str)

param_grid={'alpha':[0.001,0.01,0.1,1,10,100]}
param_grid
```

```python
grid_search=GridSearchCV(estimator=lasso, param_grid=param_grid, cv=5, scoring='r2', verbose=2)
grid_search
```

```python
grid_search.fit(x_train,y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.best_params_['alpha']
```

```python
grid_search.best_score_
```

```python
grid_search.best_estimator_
```

```python
y_pred=grid_search.best_estimator_.predict(x_test)
r2_score(y_test,y_pred)
```

```python
## Randomized search cv for lasso
```

```python
lasso=Lasso()
param_distributions={'alpha':[0.001,0.01,0.1,10,1,100]}
random_search_lasso=RandomizedSearchCV(estimator=lasso, param_distributions=param_distributions,n_iter=3, cv=5,scoring='r2',verbose=2)
random_search_lasso
```

```python
random_search_lasso.fit(x_train,y_train)
# due to n_iter parameters in each cross validation random 3 parameters are selected
```

```python
random_search_lasso.best_estimator_
```

```python
random_search_lasso.best_score_
```

```python
y_pred=random_search_lasso.best_estimator_.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# Ridge Grid Search CV
ridge=Ridge()
param_grid={'alpha':[0.001,0.01,0.1,1,10,100]}
ridge=GridSearchCV(estimator=ridge,param_grid=param_grid, cv=5, scoring='r2',verbose=2)
ridge.fit(x_train, y_train)
```

```python
ridge.best_params_
```

```python
ridge.best_estimator_
```

```python
y_pred=ridge.best_estimator_.predict(x_test)
r2_score(y_test, y_pred)
```

```python
# Ridge Randomazied search cv
ridge=Ridge()
param_distributions={'alpha':[0.001,0.01,0.1,1,10,100]}
ridge=RandomizedSearchCV(estimator=ridge,param_distributions=param_distributions,n_iter=2, cv=5, scoring='r2',verbose=2)
ridge.fit(x_train, y_train)
print(ridge.best_params_)
print(ridge.best_estimator_)
y_pred=ridge.best_estimator_.predict(x_test)
print(r2_score(y_test, y_pred))
```

### Elastic net

```python
# Elastic net 
# grid seach cv

model=ElasticNet()
param_grid={'alpha':[0.001,0.01,0.1,1,10,100],
           'l1_ratio':[0.1,0.4,0.9]}
model=GridSearchCV(estimator=model,param_grid=param_grid, cv=5, scoring='r2',verbose=2)
model.fit(x_train, y_train)
print(model.best_params_)
print(model.best_estimator_)
y_pred=model.best_estimator_.predict(x_test)
print(r2_score(y_test, y_pred))
```

```python
model=ElasticNet()
param_distributions={'alpha':[0.001,0.01,0.1,1,10,100],
                    'l1_ratio':[0.1,0.4,0.9]}
model=RandomizedSearchCV(estimator=model,param_distributions=param_distributions,n_iter=2, cv=5, scoring='r2',verbose=2)
model.fit(x_train, y_train)
print(model.best_params_)
print(model.best_estimator_)
y_pred=model.best_estimator_.predict(x_test)
print(r2_score(y_test, y_pred))
```
